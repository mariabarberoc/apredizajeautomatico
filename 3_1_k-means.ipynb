```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
```

```python
	
iris = load_iris() #cargando el dataframe iris es un diccionario con muchas cosas
iris_df = pd.DataFrame(data= iris.data, columns= iris.feature_names)

iris_df['original_label'] = iris.target #en la clusterizacion no lo vamos a hacer con la etiqueta original
iris_df.head()
```

```python
#sabemos que hay 3 clases, vamos a ir cambiando el random_state
kmeans = KMeans(n_clusters=3, max_iter=300, random_state=2) #kmeans le doy parametros 
#tengo que darle el n_clusters que es k
#mas_iter es el numero de iteraciones 
kmeans.fit(iris.data)
print(kmeans.labels_)
```

```python
kmeans = KMeans(n_clusters=3, max_iter=300, random_state=4)
kmeans.fit(iris.data)
print(kmeans.labels_)
#AQUI compruebo que sin el random cambia y asigna diferentes la entradas y los ejemplos, CAMBIA LA ASIGNACION
```

```python
from sklearn import preprocessing

cols = iris_df.columns[:-2]  # Exclude 'target' and 'cluster' columns

iris_df_norm = iris_df[cols].copy()
iris_df_norm[cols] = preprocessing.normalize(iris_df[cols])
iris_df_norm["original_label"] = iris_df["original_label"]
iris_df_norm.describe()
```

```python
	#Vemos que las flores del tipo 0 me las está metiendo todas en un cluster, lo está haciendo muy bien. Pero los otros dos, me los está dividiendo en dos, entonces no lo está ajustando pien. Esto puede ser
#por el escalado, así que vamos a probar con el reescalado.
kmeans_norm = KMeans(n_clusters=3, max_iter=300, random_state=3)
kmeans_norm.fit(iris_df_norm[cols])

iris_df_norm["cluster"] = kmeans_norm.labels_
iris_df_norm.groupby(['original_label','cluster']).agg({'sepal length (cm)': 'count'})
```

```python
kmeans_norm = KMeans(n_clusters=3, max_iter=300, random_state=2)
kmeans_norm.fit(iris_df_norm[cols])
#n_clusters=3 por eso cuando me saca la tabla en los clusters el primer cluster es 0, el segundo 1 el tercero 2 
#random_state es para iniciar en el mismo punto FIJO LA ALEATORIEDAD, 

iris_df_norm["cluster"] = kmeans_norm.labels_
iris_df_norm.groupby(['original_label','cluster']).agg({'sepal length (cm)': 'count'})
```
